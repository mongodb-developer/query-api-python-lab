# CRUD: Find

## Startup code

```python
# Import the MongoDB Driver
from pymongo import MongoClient

# Set your connection String
mongoDBURI = "mongodb://admin:mongodb@localhost:27017/?directConnection=true";

# Create a MongoDB Client
client = MongoClient(mongoDBURI)

# Get the Library Database
db = client["library"]

# Get the Books Collection
books = db["books"]
```

## Find all books containing one genre

```python
filter = { "genres": "Poetry"}

cursor = books.find(filter)

for book in cursor:
    print(f'{book["title"]} - genre: {book["genres"]}')
```

## Find all books containing the listed genres (AND)

```python
cursor = books.find(
    {"genres": {"$all": 
               ["Family Life", "Fiction"]
            }
    }).limit(10)

for book in cursor:
    print(f'{book["title"]} - genre: {book["genres"]}')

```

## Find all books containing at least one of the listed genres (OR)

```python
cursor = books.find(
    {"genres": {"$in": 
               ["Family Life", "Fiction"]
            }
    }).limit(15)

for book in cursor:
    print(f'{book["title"]} - genre: {book["genres"]}')

```

## Understanding exact array matches

When searching an array of values, MongoDB looks for an exact array match.

This means:

- the values must be the same
- the order must also match exactly

For example, the query below does not return any results because there are no documents that contain the exact array: ["Poetry, Fiction"] as the "genres" field.

```python
cursor = books.find(
    {"genres":[
        "Poetry", 
        "Fiction"
    ]}
)  

for book in cursor:
    print(f'{book["title"]}')
```
